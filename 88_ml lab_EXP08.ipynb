```python
import pandas as pd
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import MaximumLikelihoodEstimator

# Load dataset
df = pd.read_csv("heartdisease.csv")

print("Original columns:")
print(df.columns.tolist())

# Convert continuous variables into 3 categories
df['age'] = pd.cut(df['age'], bins=3, labels=['young', 'middle', 'old'])
df['trestbps'] = pd.cut(df['trestbps'], bins=3, labels=['low', 'normal', 'high'])
df['chol'] = pd.cut(df['chol'], bins=3, labels=['low', 'normal', 'high'])
df['thalach'] = pd.cut(df['thalach'], bins=3, labels=['low', 'normal', 'high'])
df['oldpeak'] = pd.cut(df['oldpeak'], bins=3, labels=['low', 'normal', 'high'])

# Convert all columns to string states
for col in df.columns:
    df[col] = df[col].astype(str)

# Create Bayesian Network
model = DiscreteBayesianNetwork([
    ('age', 'target'),
    ('sex', 'target'),
    ('cp', 'target'),
    ('trestbps', 'target'),
    ('chol', 'target'),
    ('fbs', 'target'),
    ('restecg', 'target'),
    ('thalach', 'target'),
    ('exang', 'target'),
    ('oldpeak', 'target'),
    ('slope', 'target'),
    ('ca', 'target'),
    ('thal', 'target')
])

# Maximum Likelihood Estimation
mle = MaximumLikelihoodEstimator(model, df)

# Get CPDs
cpds = mle.get_parameters()

# Add CPDs
model.add_cpds(*cpds)

# Check model
print("Model valid:", model.check_model())

# Display CPD
print(model.get_cpds('target'))
```
